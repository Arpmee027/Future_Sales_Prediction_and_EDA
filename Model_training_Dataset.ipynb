```python
import pandas as pd

# Load your data
df = pd.read_excel("Processed_Online_Retail_2.xlsx", sheet_name="Cleaned_processed_dataset")
df['Invoice_Date'] = pd.to_datetime(df['Invoice_Date'])

# Group by StockCode and Invoice_Date, sum Quantity
grouped = df.groupby(['StockCode', 'Invoice_Date'])['Quantity'].sum().reset_index()

# Create a complete date range from min to max date
all_dates = pd.date_range(start=grouped['Invoice_Date'].min(), end=grouped['Invoice_Date'].max(), freq='D')

# For each StockCode, reindex to include all dates and fill missing with 0
filled = (
    grouped.set_index('Invoice_Date')
    .groupby('StockCode')['Quantity']
    .apply(lambda x: x.reindex(all_dates, fill_value=0))
    .reset_index()
    .rename(columns={'level_1': 'Invoice_Date'})
)

# Save to Excel (multi-index or pivoted, choose one)
filled.to_excel("All_StockCodes_DailyQuantity.xlsx", index=False)
```
